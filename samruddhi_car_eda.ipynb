```python
#detect and handle outliers


plt.figure(figsize=(8,5))
plt.boxplot(df['price'])
plt.title('Boxplot of Car Price')
plt.ylabel('Price')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df['distance_travelled(kms)'])
plt.title('Boxplot of Distance Travelled')
plt.ylabel('Distance Travelled (kms)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df['car_age'])
plt.title('Boxplot of Car Age')
plt.ylabel('Car Age')
plt.show()
```

```python
for col in ['price', 'distance_travelled(kms)', 'car_age']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("\n", col)
    print("Number of outliers:", len(outliers))
```

```python
#method use
Lower Limit = Q1 − 1.5 × IQR

Upper Limit = Q3 + 1.5 × IQR
```

```python
print("Before handling outliers:", df.shape)
print("After handling outliers:", df.shape)
```
